### Step 2B: Clean up time series plots by resampling

You may notice that your plot looks a little “fuzzy”. This happens when Python is trying to plot a value for every date, but the resolution of the image is too low to actually do that. You can address this issue by **resampling** the data, or summarizing it over a time period of your choice. In this case, we will resample annually, giving us one data point per year.

```python
### find mean annual temp

ann_temp_df = (
    temp_df_units # data source
    .resample('YS') # resample frequency: annual/year start
    .mean() # aggregation method
)

ann_temp_df
```

Data observations:
* There is a discontinuity in the temperature data around 1940. 
The break in the line suggests that there is data missing for the year(s)
that are not plotted. This discontinuity matches the gap in the daily data
plotted above. Pandas should ignore the NaN values in calculating the averages, 
which means that there is likely at least one year where there is no data recorded.
This could be due to a data collection instrument being broken and no data 
being recorded or due to data loss from the source archive or other records. 

* There are a few extreme outlier average values on the plot. The most notable
are: above 20 deg C around 1940, below 7.5 deg C also around 1940, and 
below 5 deg C around 1995. As the 20 deg C spike immediately precedes the 
missing data, this could be an indication of a loss of calibration of 
instrumentation before it fails, but it could also be that the instrumentation
failed just before the cooler part of the year, so the averaged values are 
biased to warmer temperatures. The low temperature after the restart of data
suggests that that year has predominantly colder temperatures contributing
to the average. Both of these scenarios could technically be driven by an 
overall abnormally warm or cool year, but the daily data shows more white 
around those extremes, which suggests that incomplete data is the more likely cause. 
The 1995 negative spike appears more likely to be an anomalous temperature 
year as the average values around it show relatively consistent trending,
however, the daily temperature data plot shows more white space (less data)
around that time, which suggests that there may have been an instrumentation
or recording failure for the warmer part of the year, similar to the 
earlier data, which would drive the average value for the year to be 
lower than years where both extremes are captured.
